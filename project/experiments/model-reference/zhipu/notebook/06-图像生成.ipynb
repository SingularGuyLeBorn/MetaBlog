## CogView-3-Flash 模型能力

根据官方文档，CogView-3-Flash 具备以下能力：

| 能力 | 支持情况 | 说明 |
|:---|:---:|:---|
| **图像生成** | ✅ 支持 | 根据文本生成高质量图像 |
| **多分辨率** | ✅ 支持 | 支持 7 种不同分辨率 |

## 模型规格

| 规格项 | 说明 |
|:---|:---|
| 模型类型 | 图像生成模型(文生图) |

## 支持的分辨率

| 分辨率 | 比例 | 适用场景 |
|:---|:---:|:---|
| 1024x1024 | 1:1 | 方形图，社交媒体头像 |

| 864x1152 | 3:4 | 竖屏，人像 |
| 1344x768 | 16:9 | 横屏，PPT 配图 |
| 1152x864 | 4:3 | 横屏，展示图 |
| 1440x720 | 2:1 | 超宽屏，横幅 |
| 720x1440 | 1:2 | 超长图，海报 |

## ⚠️ 重要说明

> **CogView-3-Flash 是独立的图像生成模型**
> 
> - 使用独立的 API 端点 `/images/generations`，不是 `/chat/completions`
> - 不支持对话功能
> - 不支持思考模式

```python
import os
import requests
import base64
from io import BytesIO
from IPython.display import display, Image as IPImage
from openai import OpenAI

# 加载环境变量
try:
    from dotenv import load_dotenv
    load_dotenv()
    api_key = os.getenv('VITE_ZHIPU_API_KEY')
except:
    api_key = None

# 如果环境变量中没有，使用默认 API Key
if not api_key:
    api_key = "f7ec61a7856a4f4389e037eb588c69df.0pZgcFlTCk9GHCXM"

BASE_URL = "https://open.bigmodel.cn/api/paas/v4"

# 初始化客户端
client = OpenAI(
    api_key=api_key,
    base_url=BASE_URL
)

print("✅ 客户端初始化成功！")
print(f"模型：CogView-3-Flash")
print(f"类型：图像生成模型(文生图)")
print(f"API端点：/images/generations")
print(f"⚠️ 注意：不是对话模型，不能使用 chat.completions")
```

```python
# 定义图像生成函数
def generate_image(prompt, size="1024x1024", n=1, quality="standard", style="natural"):
    """
    生成图像
    
    Args:
        prompt: 图像描述文本
        size: 分辨率，支持 1024x1024, 768x1344, 864x1152, 1344x768, 1152x864, 1440x720, 720x1440
        n: 生成数量(1-4)
        quality: 质量 standard 或 hd
        style: 风格 natural 或 vivid
    
    Returns:
        生成结果
    """
    try:
        response = client.images.generate(
            model="cogview-3-flash",
            prompt=prompt,
            size=size,
            n=n,
            quality=quality,
            style=style
        )
        return response
    except Exception as e:
        print(f"❌ 生成失败：{e}")
        return None

# 定义显示图像函数
def display_image(url, width=400):
    """显示图像"""
    try:
        display(IPImage(url=url, width=width))
    except Exception as e:
        print(f"无法显示图像：{e}")
        print(f"图像URL：{url}")

print("✅ 辅助函数已定义")
```

## 2. 基础图像生成

```python
# 2.1 基础图像生成测试
print("🎨 测试1：基础图像生成\n")

prompt = "一只可爱的橘猫在草地上晒太阳，阳光明媚，背景是蓝天白云，卡通风格"

print(f"提示词：{prompt}\n")

response = generate_image(prompt, size="1024x1024")

if response and response.data:
    image_url = response.data[0].url
    print(f"✅ 图像生成成功！")
    print(f"URL：{image_url[:80]}...\n")
    print("🖼️ 生成结果：")
    display_image(image_url)
else:
    print("❌ 生成失败")
```

```python
# 2.2 不同风格测试
print("\n🎨 测试2：不同风格对比\n")

prompt = "一座现代化的城市天际线，黄昏时分"
styles = ["natural", "vivid"]

for style in styles:
    print(f"\n风格：{style}")
    print("-" * 40)
    
    response = generate_image(prompt, size="1024x1024", style=style)
    
    if response and response.data:
        display_image(response.data[0].url, width=300)
    else:
        print("生成失败")
```

## 3. 多分辨率测试

```python
# 3.1 测试不同分辨率
print("📐 测试：多分辨率支持\n")

prompt = "一片美丽的樱花树林，春天，粉色花瓣飘落"

# 常用分辨率
sizes = [
    ("1024x1024", "1:1 方形"),
    ("1344x768", "16:9 横屏"),
    ("768x1344", "9:16 竖屏"),
]

for size, desc in sizes:
    print(f"\n分辨率：{size} ({desc})")
    print("=" * 50)
    
    response = generate_image(prompt, size=size)
    
    if response and response.data:
        # 根据方向调整显示宽度
        width = 300 if "1024" in size else (400 if "1344" in size and "768" in size else 250)
        display_image(response.data[0].url, width=width)
    else:
        print("生成失败")
```

## 4. 批量生成

```python
# 4.1 批量生成测试
print("🎨 测试：批量生成(一次生成多张)\n")

prompt = "一只可爱的小狗在公园里玩耍，写实风格"

print(f"提示词：{prompt}")
print(f"生成数量：4 张\n")

response = generate_image(prompt, size="1024x1024", n=4)

if response and response.data:
    print(f"✅ 成功生成 {len(response.data)} 张图像\n")
    
    for i, img_data in enumerate(response.data, 1):
        print(f"图像 {i}：")
        display_image(img_data.url, width=250)
        print()
else:
    print("❌ 生成失败")
```

## 9. 错误处理测试

```python
# 9.1 测试：不支持的分辨率
print("⚠️ 测试：不支持的分辨率\n")

try:
    response = generate_image("一只猫", size="1920x1080")  # 不支持的分辨率
    if response:
        print("⚠️ 可能使用了默认分辨率")
        display_image(response.data[0].url, width=300)
except Exception as e:
    print(f"✅ 正确捕获错误：{type(e).__name__}")
    print(f"   错误：{str(e)[:100]}...")
```

```python
# 9.2 测试：空提示词
print("\n⚠️ 测试：空提示词\n")

try:
    response = generate_image("")
    if response:
        print("⚠️ 响应结果：")
        print(response)
except Exception as e:
    print(f"✅ 错误捕获：{type(e).__name__}")
    print(f"   错误：{str(e)[:100]}...")
```

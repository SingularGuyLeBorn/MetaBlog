## 2. GLM-4.6V-Flash 视频理解基础测试

测试基本的视频理解能力，让模型描述视频内容。

GLM-4.6V-Flash 支持思考模式，可通过 `thinking: {"type": "enabled"}` 启用。

```python
def test_basic_video_understanding():
    """测试基础视频理解"""
    print("\n🎬 测试1: 基础视频理解")
    print("="*50)
    
    try:
        response = client.chat.completions.create(
            model="glm-4.6v-flash",
            messages=[
                {
                    "role": "user",
                    "content": [
                        {
                            "type": "video_url",
                            "video_url": {
                                "url": TEST_VIDEO_URL
                            }
                        },
                        {
                            "type": "text",
                            "text": "请描述这个视频的内容"
                        }
                    ]
                }
            ]
        )
        
        result = response.choices[0].message.content
        print("\n✅ 成功响应:")
        display(Markdown(result))
        
        # 显示 Token 使用情况
        usage = response.usage
        print(f"\n📊 Token 使用: 输入 {usage.prompt_tokens}, 输出 {usage.completion_tokens}, 总计 {usage.total_tokens}")
        
        return True
        
    except Exception as e:
        print(f"\n❌ 错误: {e}")
        return False

test_basic_video_understanding()
```

## 6. 视频 + 文本混合输入测试

测试视频与文本的混合输入能力。

```python
def test_mixed_input():
    """测试视频+文本混合输入"""
    print("\n🔄 测试5: 视频+文本混合输入")
    print("="*50)
    
    test_cases = [
        {
            "name": "视频+背景信息",
            "content": [
                {
                    "type": "text",
                    "text": "这是一段演示视频"
                },
                {
                    "type": "video_url",
                    "video_url": {
                        "url": TEST_VIDEO_URL
                    }
                },
                {
                    "type": "text",
                    "text": "请根据以上背景信息描述视频内容"
                }
            ]
        },
        {
            "name": "视频+具体问题",
            "content": [
                {
                    "type": "video_url",
                    "video_url": {
                        "url": TEST_VIDEO_URL
                    }
                },
                {
                    "type": "text",
                    "text": "基于以上视频，请回答：1) 这是什么场景？ 2) 主要活动是什么？"
                }
            ]
        },
        {
            "name": "视频+分析要求",
            "content": [
                {
                    "type": "video_url",
                    "video_url": {
                        "url": TEST_VIDEO_URL
                    }
                },
                {
                    "type": "text",
                    "text": "请以结构化的方式分析这个视频，包括：场景、人物、动作、物品、时间线索"
                }
            ]
        }
    ]
    
    for test in test_cases:
        print(f"\n--- {test['name']} ---")
        
        try:
            response = client.chat.completions.create(
                model="glm-4.6v-flash",
                messages=[
                    {
                        "role": "user",
                        "content": test["content"]
                    }
                ]
            )
            
            result = response.choices[0].message.content
            display(Markdown(result))
            
        except Exception as e:
            print(f"❌ 错误: {e}")

test_mixed_input()
```

## 7. 错误处理测试

测试各种错误情况的处理。

```python
def test_error_handling():
    """测试错误处理"""
    print("\n⚠️ 测试6: 错误处理")
    print("="*50)
    
    # 测试用例
    error_tests = [
        {
            "name": "无效的视频URL",
            "video_url": "https://invalid-url.com/video.mp4",
            "prompt": "请描述这个视频"
        },
        {
            "name": "不存在的视频",
            "video_url": "https://cdn.bigmodel.cn/nonexistent.mp4",
            "prompt": "请描述这个视频"
        },
        {
            "name": "错误的模型",
            "video_url": TEST_VIDEO_URL,
            "prompt": "请描述这个视频",
            "model": "glm-4v-flash"  # 错误的模型，不支持视频
        },
        {
            "name": "空提示词",
            "video_url": TEST_VIDEO_URL,
            "prompt": ""
        }
    ]
    
    for test in error_tests:
        print(f"\n--- 测试: {test['name']} ---")
        print(f"视频URL: {test['video_url'][:50]}...")
        
        try:
            model = test.get("model", "glm-4.6v-flash")
            response = client.chat.completions.create(
                model=model,
                messages=[
                    {
                        "role": "user",
                        "content": [
                            {
                                "type": "video_url",
                                "video_url": {
                                    "url": test["video_url"]
                                }
                            },
                            {
                                "type": "text",
                                "text": test["prompt"] or "请描述"
                            }
                        ]
                    }
                ]
            )
            
            result = response.choices[0].message.content
            print(f"✅ 响应成功:")
            print(result[:200] + "..." if len(result) > 200 else result)
            
        except Exception as e:
            print(f"❌ 错误类型: {type(e).__name__}")
            print(f"❌ 错误信息: {e}")

test_error_handling()
```

## 8. 使用原生 HTTP 请求测试

使用 requests 库直接发送 HTTP 请求进行测试。

```python
def test_raw_http_request():
    """使用原生 HTTP 请求测试"""
    print("\n🌐 测试7: 原生 HTTP 请求")
    print("="*50)
    
    headers = {
        "Authorization": f"Bearer {API_KEY}",
        "Content-Type": "application/json"
    }
    
    payload = {
        "model": "glm-4.6v-flash",
        "messages": [
            {
                "role": "user",
                "content": [
                    {
                        "type": "video_url",
                        "video_url": {
                            "url": TEST_VIDEO_URL
                        }
                    },
                    {
                        "type": "text",
                        "text": "请用一句话概括这个视频"
                    }
                ]
            }
        ]
    }
    
    try:
        response = requests.post(
            f"{BASE_URL}/chat/completions",
            headers=headers,
            json=payload,
            timeout=120
        )
        
        print(f"状态码: {response.status_code}")
        
        if response.status_code == 200:
            result = response.json()
            content = result["choices"][0]["message"]["content"]
            print("\n✅ 成功响应:")
            display(Markdown(content))
            
            # 显示完整响应结构
            print("\n📋 完整响应结构:")
            print(json.dumps(result, indent=2, ensure_ascii=False)[:1000] + "...")
        else:
            print(f"❌ 请求失败: {response.text}")
            
    except Exception as e:
        print(f"❌ 错误: {e}")

test_raw_http_request()
```

# 03-深度思考(思维链)测试

## GLM-4.1V-Thinking-Flash 模型能力

根据官方文档，GLM-4.1V-Thinking-Flash 具备以下能力：

| 能力 | 支持情况 | 说明 |
|:---|:---:|:---|
| **内置深度思考** | ✅ 内置 | 默认内置，**强制思考**，始终返回思维链 |
| **视觉理解** | ✅ 支持 | 支持图片、视频、文件 |
| **流式输出** | ✅ 支持 | 实时流式响应 |

## 模型规格

| 规格项 | 数值 |
|:---|:---|
| 输入模态 | 图像、视频、文件、文本 |
| 输出模态 | 文本 |
| 上下文窗口 | 64K |
| 思考模式 | **强制思考**(不可关闭) |

## ⚠️ 关键区别

与 GLM-4.6V-Flash 的区别：

| 模型 | 思考模式 | 控制方式 |
|:---|:---|:---|
| GLM-4.1V-Thinking-Flash | **内置** | 强制思考，始终返回 reasoning_content |
| GLM-4.6V-Flash | **可选** | 可通过 `thinking` 参数开启/关闭 |

## 功能特色

1. **图文理解** - 精准识别并综合分析图像与文本信息
2. **数学与科学推理** - 支持复杂题解、多步演绎与公式理解
3. **视频理解** - 具备时序分析与事件逻辑建模能力
4. **GUI 与网页智能体任务** - 理解界面结构，辅助自动化操作
5. **视觉锚定与实体定位** - 语言与图像区域精准对齐

## 2. 图片问答(图表分析)

```python
# 2.1 图表分析 - 找出年度利润最高的公司
print("📊 测试：图表分析 - 公司利润分析\n")
print("问题：请找出这张图中年度利润最高的公司，以及该公司的最大部门？\n")

response = client.chat.completions.create(
    model="glm-4.1v-thinking-flash",
    messages=[
        {
            "role": "user",
            "content": [
                {
                    "type": "image_url",
                    "image_url": {"url": CHART_IMAGE}
                },
                {
                    "type": "text",
                    "text": "请找出这张图中年度利润最高的公司，以及该公司的最大部门？"
                }
            ]
        }
    ],
    max_tokens=2048
)

message = response.choices[0].message

# 显示思维链
if hasattr(message, 'reasoning_content') and message.reasoning_content:
    print("🧠 思维链(推理过程)：")
    print("-" * 60)
    print(message.reasoning_content)
    print("-" * 60)
    print("\n")

print("💡 最终答案：")
print(message.content)
```

## 7. 流式输出 + 思维链

```python
# 7.1 流式输出测试
print("🌊 测试：流式输出 + 思维链\n")

response = client.chat.completions.create(
    model="glm-4.1v-thinking-flash",
    messages=[
        {
            "role": "user",
            "content": [
                {
                    "type": "image_url",
                    "image_url": {"url": CHART_IMAGE}
                },
                {
                    "type": "text",
                    "text": "分析这张图表的主要趋势"
                }
            ]
        }
    ],
    stream=True,
    max_tokens=2048
)

print("🧠 思维链：")
print("-" * 60)
reasoning_parts = []
content_parts = []

for chunk in response:
    delta = chunk.choices[0].delta
    
    # 获取思维链
    if hasattr(delta, 'reasoning_content') and delta.reasoning_content:
        print(delta.reasoning_content, end="", flush=True)
        reasoning_parts.append(delta.reasoning_content)
    
    # 获取最终内容
    if delta.content:
        content_parts.append(delta.content)

print("\n" + "-" * 60)
print("\n💡 最终分析：")
print("".join(content_parts))
```

## 8. 对比测试：GLM-4.6V-Flash vs GLM-4.1V-Thinking-Flash

```python
# 8.1 对比两个模型在相同任务上的表现
print("📊 对比测试：两个视觉模型的推理能力\n")

test_question = "这张图表展示了几家公司的什么数据？数据之间有什么关联？"

models = [
    ("GLM-4.6V-Flash (无思考)", "glm-4.6v-flash", {"thinking": {"type": "disabled"}}),
    ("GLM-4.6V-Flash (有思考)", "glm-4.6v-flash", {"thinking": {"type": "enabled"}}),
    ("GLM-4.1V-Thinking-Flash (强制思考)", "glm-4.1v-thinking-flash", {})
]

for name, model_code, extra_params in models:
    print(f"\n{'='*60}")
    print(f"🤖 模型：{name}")
    print(f"{'='*60}\n")
    
    try:
        response = client.chat.completions.create(
            model=model_code,
            messages=[
                {
                    "role": "user",
                    "content": [
                        {"type": "image_url", "image_url": {"url": CHART_IMAGE}},
                        {"type": "text", "text": test_question}
                    ]
                }
            ],
            max_tokens=2048,
            **extra_params
        )
        
        message = response.choices[0].message
        
        # 检查是否有思维链
        has_reasoning = hasattr(message, 'reasoning_content') and message.reasoning_content
        
        if has_reasoning:
            print(f"🧠 思维链长度：{len(message.reasoning_content)} 字符")
            print(f"💡 答案长度：{len(message.content)} 字符")
        else:
            print(f"❌ 无思维链")
            print(f"💡 答案长度：{len(message.content)} 字符")
        
        print(f"\n回答预览：{message.content[:200]}...")
        
    except Exception as e:
        print(f"❌ 错误：{e}")

print(f"\n{'='*60}")
print("📊 对比总结：")
print("• GLM-4.6V-Flash (无思考)：直接回答，无推理过程")
print("• GLM-4.6V-Flash (有思考)：可选开启思维链")
print("• GLM-4.1V-Thinking-Flash：强制思考，始终返回思维链")
```

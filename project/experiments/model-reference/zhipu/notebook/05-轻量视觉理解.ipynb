# 05-轻量视觉理解测试

## GLM-4V-Flash 模型能力

根据官方文档，GLM-4V-Flash 具备以下能力：

| 能力 | 支持情况 | 说明 |
|:---|:---:|:---|
| **视觉理解** | ✅ 支持 | 支持图片理解 |
| **流式输出** | ✅ 支持 | 实时流式响应 |
| **思考模式** | ❌ 不支持 | 该模型不支持深度思考 |
| **工具调用** | ❌ 不支持 | 该模型不支持 Function Call |

## 模型规格

| 规格项 | 数值 |
|:---|:---|
| 输入模态 | 图片、文本 |
| 输出模态 | 文本 |
| 上下文窗口 | 16K |
| 最大输出 Tokens | 1K |
| 图片数量限制 | **仅支持 1 张** |
| Base64 支持 | **不支持** |

## ⚠️ 重要限制

> **仅支持单张图片，且不支持 Base64 编码**
> 
> **不支持视频和文件理解**(仅 GLM-4.6V-Flash 和 GLM-4.1V-Thinking-Flash 支持)

## 功能特色

1. **图像描述生成** - 自动生成图像的文字描述
2. **图像分类** - 对图像进行分类识别
3. **图像情感分析** - 分析图像传达的情感

## 场景应用

- 图表问答 - 分析图表数据趋势
- 社交媒体内容生成 - 根据图片生成文案
- 教育应用 - 解释图像中的科学现象
- 质量检测 - 识别图像中的缺陷
- 商品描述生成 - 自动生成商品标题
- 肤质分析 - 提供护肤建议

## 2. 图像描述生成

```python
# 2.1 基础图像描述
print("🖼️ 测试：图像描述生成\n")

response = client.chat.completions.create(
    model="glm-4v-flash",
    messages=[
        {
            "role": "user",
            "content": [
                {
                    "type": "image_url",
                    "image_url": {"url": TEST_IMAGE}
                },
                {
                    "type": "text",
                    "text": "请详细描述这张图片的内容"
                }
            ]
        }
    ],
    max_tokens=1024
)

print("🤖 图像描述：")
print(response.choices[0].message.content)
print(f"\n📊 Token使用：{response.usage.total_tokens}")
```

## 10. 流式输出测试

```python
# 10.1 流式输出
print("🌊 测试：流式输出\n")
print("🤖 模型回复：", end="", flush=True)

response = client.chat.completions.create(
    model="glm-4v-flash",
    messages=[
        {
            "role": "user",
            "content": [
                {
                    "type": "image_url",
                    "image_url": {"url": TEST_IMAGE}
                },
                {
                    "type": "text",
                    "text": "描述这张图片"
                }
            ]
        }
    ],
    stream=True,
    max_tokens=512
)

for chunk in response:
    if chunk.choices[0].delta.content:
        print(chunk.choices[0].delta.content, end="", flush=True)

print("\n\n✅ 流式输出完成")
```

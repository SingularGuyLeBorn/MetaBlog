# 05-图像理解 - Kimi API

本文档演示 Kimi K2.5 的图像理解能力。

```python
from openai import OpenAI
import os
import base64
from dotenv import load_dotenv

load_dotenv(dotenv_path='../../.env')

api_key = os.getenv("VITE_KIMI_API_KEY") or os.getenv("KIMI_API_KEY")
base_url = os.getenv("VITE_KIMI_BASE_URL", "https://api.moonshot.cn/v1")

client = OpenAI(api_key=api_key, base_url=base_url)
print("✅ Kimi 客户端初始化成功")
```

## 图片编码辅助函数

```python
def encode_image(image_path: str) -> str:
    """将图片编码为 base64"""
    with open(image_path, "rb") as f:
        ext = os.path.splitext(image_path)[1].lower().replace(".", "")
        if ext == "jpg":
            ext = "jpeg"
        base64_data = base64.b64encode(f.read()).decode("utf-8")
        return f"data:image/{ext};base64,{base64_data}"

# 注意：运行以下示例前，请确保有示例图片
# image_path = "example.jpg"  # 替换为实际图片路径
```

## 基础图像理解

```python
# 基础图像理解
image_path = "test_ocr.png"
image_url = encode_image(image_path)

response = client.chat.completions.create(
    model="kimi-k2.5",
    messages=[
        {
            "role": "user",
            "content": [
                {"type": "image_url", "image_url": {"url": image_url}},
                {"type": "text", "text": "请描述这张图片的内容。"}
            ]
        }
    ],
)

print("📄 图像描述:")
print(response.choices[0].message.content)
```

## 远程图片通过 file_id 给 Kimi 看（微信文章图片场景）

测试目标：验证 `ms://file_id` 方式能否让 Kimi 看到微信文章中的图片原图。

流程：
1. 下载微信图片（带 Referer 绕过防盗链）
2. 上传到 Kimi API 获取 `file_id`
3. 用 `ms://file_id` 调用 vision 模型

**测试结果：成功。**

```python
import os
import requests
from openai import OpenAI
from dotenv import load_dotenv

load_dotenv(dotenv_path="../../../../../.env")

api_key = os.getenv("VITE_KIMI_API_KEY") or os.getenv("KIMI_API_KEY")
base_url = os.getenv("VITE_KIMI_BASE_URL", "https://api.moonshot.cn/v1")

client = OpenAI(api_key=api_key, base_url=base_url)

# 微信图片链接
IMAGE_URL = "https://mmbiz.qpic.cn/mmbiz_jpg/7T8j2pUqCWsjy18EqR9vriaL63DOLf3WxgJQY0Ac65ibcIeJjGHG2WHmm8S4Q1IRInAicUo6eu5ibd1tqAgtHLnoKG6LbUZfG6uZGRsQ9NKczTw/0?wx_fmt=jpeg"
TEMP_PATH = "test_wechat_image.jpg"

print("步骤 1: 下载微信图片")
headers = {
    "User-Agent": "Mozilla/5.0 ...",
    "Referer": "https://mp.weixin.qq.com/",
}

resp = requests.get(IMAGE_URL, headers=headers, timeout=30)
print(f"状态: {resp.status_code}, 大小: {len(resp.content)} bytes")

with open(TEMP_PATH, "wb") as f:
    f.write(resp.content)

print("步骤 2: 上传到 Kimi")
file_obj = client.files.create(file=open(TEMP_PATH, "rb"), purpose="image")
print(f"file_id: {file_obj.id}, status: {file_obj.status}")

print("步骤 3: 用 ms://file_id 调用 vision")
response = client.chat.completions.create(
    model="kimi-k2.5",
    messages=[
        {
            "role": "user",
            "content": [
                {"type": "image_url", "image_url": {"url": f"ms://{file_obj.id}"}},
                {"type": "text", "text": "请描述这张图片的内容"}
            ]
        }
    ],
)

print("\nKimi 回复:")
print(response.choices[0].message.content)

os.remove(TEMP_PATH)
print(f"\n已清理: {TEMP_PATH}")
```
